```python
import pandas as pd

# Caricamento del file CSV 
df = pd.read_csv('../../resorts_by_country/Denmark.csv')

# Dizionario da compilare manualmente: "nome località" → (latitudine, longitudine)
coordinate_dict = {
    "Dry slopes Dayz Søhøjlandets Skicenter": (56.1636, 9.7076), 
    "Indoor ski area Skidome Denmark – Randers (planned)": (56.4604, 10.0363), 
    "Dry slopes Indoorski – Rødovre": (55.4053, 12.2718), 
    "Dry slopes Ski Arena – Aarhus": (56.1174, 10.1193), 
    "Copenhagen (planned)": (55.4034, 12.3406), 
    "Østerlars (Bornholm)": (55.0953, 14.5741), 
    "Hoch Hylkedal – Kolding": (55.4959, 9.4730), 
    "Hedelands Skicenter": (55.6271, 12.1756)
}

# Funzione per recuperare lat e lon dal dizionario
def get_lat(place):
    return coordinate_dict.get(place, (None, None))[0]

def get_lon(place):
    return coordinate_dict.get(place, (None, None))[1]

# Applica le funzioni
df['latitude'] = df['Resort'].apply(get_lat)
df['longitude'] = df['Resort'].apply(get_lon)

# Salva il nuovo file con le coordinate aggiunte
df.to_csv('../../resorts_by_country/coordinates/DK_with_coordinates.csv', index=False)

print("File salvato come 'DK_with_coordinates.csv'")
```
