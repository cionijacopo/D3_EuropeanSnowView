```python
import pandas as pd

# Caricamento del file CSV 
df = pd.read_csv('../../resorts_by_country/Slovakia.csv')

# Dizionario da compilare manualmente: "nome località" → (latitudine, longitudine)
coordinate_dict = {
    "Jasná Nízke Tatry-Chopok": (48.5814, 19.3505), 
    "Donovaly-Park Snow-": (48.5241, 19.1411), 
    "Štrbské Pleso": (49.0721, 20.0330),
    "Kubínska hoľa": (49.1352, 19.1826), 
    "Tatranská Lomnica": (49.0953, 20.1646)
}

# Funzione per recuperare lat e lon dal dizionario
def get_lat(place):
    return coordinate_dict.get(place, (None, None))[0]

def get_lon(place):
    return coordinate_dict.get(place, (None, None))[1]

# Applica le funzioni
df['latitude'] = df['Resort'].apply(get_lat)
df['longitude'] = df['Resort'].apply(get_lon)

# Salva il nuovo file con le coordinate aggiunte
df.to_csv('../../resorts_by_country/coordinates/SK_with_coordinates.csv', index=False)

print("File salvato come 'SK_with_coordinates.csv'")
```
