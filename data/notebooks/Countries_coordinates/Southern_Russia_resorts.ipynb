```python
import pandas as pd

# Caricamento del file CSV 
df = pd.read_csv('../../resorts_by_country/Southern_Russia.csv')

# Dizionario da compilare manualmente: "nome località" → (latitudine, longitudine)
coordinate_dict = {
    "Rosa Khutor": (44.62, 39.29), 
    "Gornaya Karusel": (44.7029, 40.2667), 
    "Gazprom Mountain Resort-Laura-– Esto-Sadok-Sotschi-": (43.6774, 39.2809), 
    "Alpika Service": (43.72, 40.20)
}

# Funzione per recuperare lat e lon dal dizionario
def get_lat(place):
    return coordinate_dict.get(place, (None, None))[0]

def get_lon(place):
    return coordinate_dict.get(place, (None, None))[1]

# Applica le funzioni
df['latitude'] = df['Resort'].apply(get_lat)
df['longitude'] = df['Resort'].apply(get_lon)

# Salva il nuovo file con le coordinate aggiunte
df.to_csv('../../resorts_by_country/coordinates/RU_with_coordinates.csv', index=False)

print("File salvato come 'RU_with_coordinates.csv'")
```
