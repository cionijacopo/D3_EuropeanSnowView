```python
import pandas as pd

# Caricamento del file CSV 
df = pd.read_csv('../../resorts_by_country/Bulgaria.csv')

# Dizionario da compilare manualmente: "nome località" → (latitudine, longitudine)
coordinate_dict = {
    "Vitosha-Sofia": (42.6367, 23.2438), 
    "Borovets": (42.7, 23.0), 
    "Pamporovo": (41.6567, 24.6953), 
    "Bansko": (41.8404, 23.4856)
}

# Funzione per recuperare lat e lon dal dizionario
def get_lat(place):
    return coordinate_dict.get(place, (None, None))[0]

def get_lon(place):
    return coordinate_dict.get(place, (None, None))[1]

# Applica le funzioni
df['latitude'] = df['Resort'].apply(get_lat)
df['longitude'] = df['Resort'].apply(get_lon)

# Salva il nuovo file con le coordinate aggiunte
df.to_csv('../../resorts_by_country/coordinates/BG_with_coordinates.csv', index=False)

print("File salvato come 'BG_with_coordinates.csv'")
```
