```python
import pandas as pd

# Caricamento del file CSV 
df = pd.read_csv('../../resorts_by_country/United_Kingdom.csv')

# Dizionario da compilare manualmente: "nome località" → (latitudine, longitudine)
coordinate_dict = {
    "Glenshee": (56.8873, -3.4157),
    "Glencoe Mountain": (56.6258, -4.8225),
    "The Lecht": (57.1929, -3.2396), 
    "Nevis Range": (56.8124, -4.9618)
}

# Funzione per recuperare lat e lon dal dizionario
def get_lat(place):
    return coordinate_dict.get(place, (None, None))[0]

def get_lon(place):
    return coordinate_dict.get(place, (None, None))[1]

# Applica le funzioni
df['latitude'] = df['Resort'].apply(get_lat)
df['longitude'] = df['Resort'].apply(get_lon)

# Salva il nuovo file con le coordinate aggiunte
df.to_csv('../../resorts_by_country/coordinates/GB_with_coordinates.csv', index=False)

print("File salvato come 'GB_with_coordinates.csv'")
```
