```python
import pandas as pd

# Caricamento del file CSV 
df = pd.read_csv('../../resorts_by_country/Sweden.csv')

# Dizionario da compilare manualmente: "nome località" → (latitudine, longitudine)
coordinate_dict = {
    "Åre": (63.2337, 12.5606), 
    "Lindvallen-​Högfjället-Sälen-": (61.0924, 13.0701), 
    "Tandådalen-​Hundfjället-Sälen-": (61.16, 12.98), 
    "Stöten": (61.1555, 12.5330),
    "Idre Fjäll": (61.5322, 12.5002), 
    "Tänndalen": (62.5436, 12.3325)
}

# Funzione per recuperare lat e lon dal dizionario
def get_lat(place):
    return coordinate_dict.get(place, (None, None))[0]

def get_lon(place):
    return coordinate_dict.get(place, (None, None))[1]

# Applica le funzioni
df['latitude'] = df['Resort'].apply(get_lat)
df['longitude'] = df['Resort'].apply(get_lon)

# Salva il nuovo file con le coordinate aggiunte
df.to_csv('../../resorts_by_country/coordinates/SE_with_coordinates.csv', index=False)

print("File salvato come 'SE_with_coordinates.csv'")
```
