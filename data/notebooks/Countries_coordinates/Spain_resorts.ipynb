```python
import pandas as pd

# Caricamento del file CSV 
df = pd.read_csv('../../resorts_by_country/Spain.csv')

# Dizionario da compilare manualmente: "nome località" → (latitudine, longitudine)
coordinate_dict = {
    "La Molina-​Masella-Alp2500": (42.2037, 1.5722), 
    "Cerler": (42.59, 0.54), 
    "Candanchú": (42.4717, 0.5940), 
    "Vall de Núria": (42.2359, 2.0860), 
    "Port del Comte": (42.1022, 1.3343), 
    "Formigal": (42.4626, 0.2139), 
    "Baqueira / ​Beret": (42.4157, 0.5648), 
    "Sierra Nevada-Pradollano": (37.09, -3.44)
}

# Funzione per recuperare lat e lon dal dizionario
def get_lat(place):
    return coordinate_dict.get(place, (None, None))[0]

def get_lon(place):
    return coordinate_dict.get(place, (None, None))[1]

# Applica le funzioni
df['latitude'] = df['Resort'].apply(get_lat)
df['longitude'] = df['Resort'].apply(get_lon)

# Salva il nuovo file con le coordinate aggiunte
df.to_csv('../../resorts_by_country/coordinates/ES_with_coordinates.csv', index=False)

print("File salvato come 'ES_with_coordinates.csv'")
```
