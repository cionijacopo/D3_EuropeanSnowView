```python
import pandas as pd

# Caricamento del file CSV 
df = pd.read_csv('../../resorts_by_country/Germany.csv')

# Dizionario da compilare manualmente: "nome località" → (latitudine, longitudine)
coordinate_dict = {
    "Buchenberg-Buching-Halblech-": (47.3622, 10.4840), 
    "Zugspitze": (47.2516, 10.5907), 
    "Garmisch-Classic-Garmisch-Partenkirchen": (47.2816, 11.0350), 
    "Wendelstein-Brannenburg-​Osterhofen": (47.4210, 12.0044), 
    "Feldberg-Seebuck-​Grafenmatt-​Fahl": (47.8597, 8.0363), 
    "Mitterdorf-Almberg-– Mitterfirmiansreut": (48.858, 13.542), 
    "Geißkopf-Bischofsmais": (48.9241, 13.0372), 
    "Indoor ski area Snow Dome Bispingen": (53.065, 9.5828), 
    "Sudelfeld-Bayrischzell": (47.4016, 12.0224),
    "Grüntenlifte-Kranzegg-Rettenberg-": (47.58, 10.30), 
    "Nebelhorn-Oberstdorf": (47.2518, 10.2033), 
    "Eibenstock": (50.2944, 12.3551), 
    "Fichtelberg-Oberwiesenthal": (50.2543, 12.5717), 
    "Schöneck-Skiwelt-": (50.2322, 12.2047), 
    "Silbersattel-Steinach": (50.2610, 11.0757), 
    "Grasgehren-Bolgengrat": (47.44, 10.18),
    "Götschen-Bischofswiesen": (47.63, 12.93), 
    "Unternberg-Ruhpolding-": (47.73, 12.63), 
    "Postwiesen Skidorf-Neuastenberg": (51.0959, 8.2911), 
    "Oberjoch-Bad Hindelang-– Iseler": (47.2955, 10.2504), 
    "Kranzberg-Mittenwald": (47.45, 11.23), 
    "Rossfeld - Berchtesgaden - Oberau": (47.39, 13.0337), 
    "Brauneck Lenggries ​Wegscheid": (47.3949, 11.3146), 
    "Hochschwarzeck": (47.3737, 12.5418)
}

# Funzione per recuperare lat e lon dal dizionario
def get_lat(place):
    return coordinate_dict.get(place, (None, None))[0]

def get_lon(place):
    return coordinate_dict.get(place, (None, None))[1]

# Applica le funzioni
df['latitude'] = df['Resort'].apply(get_lat)
df['longitude'] = df['Resort'].apply(get_lon)

# Salva il nuovo file con le coordinate aggiunte
df.to_csv('../../resorts_by_country/coordinates/DE_with_coordinates.csv', index=False)

print("File salvato come 'DE_with_coordinates.csv'")
```
