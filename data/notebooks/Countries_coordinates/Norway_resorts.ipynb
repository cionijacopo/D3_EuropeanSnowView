```python
import pandas as pd

# Caricamento del file CSV 
df = pd.read_csv('../../resorts_by_country/Norway.csv')

# Dizionario da compilare manualmente: "nome località" → (latitudine, longitudine)
coordinate_dict = {
    "Trysil": (61.1836, 12.1854), 
    "Oppdal": (62.34, 9.36), 
    "Myrkdalen": (60.8421, 6.4654), 
    "Gålå": (61.5057, 9.7848), 
    "Kvitfjell": (61.45, 10.12), 
    "Hafjell": (61.1432, 10.2631), 
    "Voss": (60.4209, 06.2523), 
    "Geilosiden Geilo": (60.57, 8.23), 
    "Fonna Glacier": (60.00, 6.20), 
    "Hemsedal": (60.5421, 8.3053)
}

# Funzione per recuperare lat e lon dal dizionario
def get_lat(place):
    return coordinate_dict.get(place, (None, None))[0]

def get_lon(place):
    return coordinate_dict.get(place, (None, None))[1]

# Applica le funzioni
df['latitude'] = df['Resort'].apply(get_lat)
df['longitude'] = df['Resort'].apply(get_lon)

# Salva il nuovo file con le coordinate aggiunte
df.to_csv('../../resorts_by_country/coordinates/NO_with_coordinates.csv', index=False)

print("File salvato come 'NO_with_coordinates.csv'")
```
