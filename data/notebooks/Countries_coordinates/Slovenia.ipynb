```python
import pandas as pd

# Caricamento del file CSV 
df = pd.read_csv('../../resorts_by_country/Slovenia.csv')

# Dizionario da compilare manualmente: "nome località" → (latitudine, longitudine)
coordinate_dict = {
    "Kobla-Bohinj": (46.2674, 13.9587), 
    "Mariborsko Pohorje-Maribor": (46.3013, 15.2711), 
    "Kanin-Sella Nevea-​Bovec": (46.3376, 13.2832), 
    "Kranjska Gora": (46.2907, 13.4710)
}

# Funzione per recuperare lat e lon dal dizionario
def get_lat(place):
    return coordinate_dict.get(place, (None, None))[0]

def get_lon(place):
    return coordinate_dict.get(place, (None, None))[1]

# Applica le funzioni
df['latitude'] = df['Resort'].apply(get_lat)
df['longitude'] = df['Resort'].apply(get_lon)

# Salva il nuovo file con le coordinate aggiunte
df.to_csv('../../resorts_by_country/coordinates/SI_with_coordinates.csv', index=False)

print("File salvato come 'SI_with_coordinates.csv'")
```
