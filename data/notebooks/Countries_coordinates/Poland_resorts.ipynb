```python
import pandas as pd

# Caricamento del file CSV 
df = pd.read_csv('../../resorts_by_country/Poland.csv')

# Dizionario da compilare manualmente: "nome località" → (latitudine, longitudine)
coordinate_dict = {
    "Białka Tatrzańska-Kotelnica-​Kaniówka-​Bania": (49.3587, 20.1577), 
    "Szczyrk-Skrzyczne": (49.4104, 19.0148)
}

# Funzione per recuperare lat e lon dal dizionario
def get_lat(place):
    return coordinate_dict.get(place, (None, None))[0]

def get_lon(place):
    return coordinate_dict.get(place, (None, None))[1]

# Applica le funzioni
df['latitude'] = df['Resort'].apply(get_lat)
df['longitude'] = df['Resort'].apply(get_lon)

# Salva il nuovo file con le coordinate aggiunte
df.to_csv('../../resorts_by_country/coordinates/PL_with_coordinates.csv', index=False)

print("File salvato come 'PL_with_coordinates.csv'")
```
