```python
import pandas as pd

# Caricamento del file CSV 
df = pd.read_csv('../../resorts_by_country/Andorra.csv')

# Dizionario da compilare manualmente: "nome località" → (latitudine, longitudine)
coordinate_dict = {
    "Soldeu-Pas de la Casa/​Grau Roig/​El Tarter/​Canillo/​Encamp (Grandvalira)": (42.58, 1.67), 
    "Pal-​Arinsal-La Massana-Vallnord-": (42.34, 1.29), 
    "Grandvalira – Pas de la Casa/​Grau Roig/​Soldeu/​El Tarter/​Canillo/​Encamp": (42.3232, 1.4401), 
    "El Tarter – Pas de la Casa/​Grau Roig/​Soldeu/​Canillo/​Encamp (Grandvalira)": (42.58, 1.67), 
    "Arcalís-Ordino (Vallnord)": (42.3755, 1.2958)
}

# Funzione per recuperare lat e lon dal dizionario
def get_lat(place):
    return coordinate_dict.get(place, (None, None))[0]

def get_lon(place):
    return coordinate_dict.get(place, (None, None))[1]

# Applica le funzioni
df['latitude'] = df['Resort'].apply(get_lat)
df['longitude'] = df['Resort'].apply(get_lon)

# Salva il nuovo file con le coordinate aggiunte
df.to_csv('../../resorts_by_country/coordinates/AD_with_coordinates.csv', index=False)

print("File salvato come 'AD_with_coordinates.csv'")
```
